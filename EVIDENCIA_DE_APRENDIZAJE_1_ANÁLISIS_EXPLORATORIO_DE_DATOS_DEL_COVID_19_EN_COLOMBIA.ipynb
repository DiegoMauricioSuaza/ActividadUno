Análisis Descriptivo de Datos (ADD)

1- Realiza un análisis descriptivo de datos para todas las variables

```python
# Valores faltantes por variable
faltantes = data.isnull().sum()

print(faltantes)
```

```python
valores_unicos = pd.DataFrame({
    'Variable': data.columns,
    'Valores únicos': [data[col].nunique(dropna=False) for col in data.columns]
})

valores_unicos
```

Análisis de la variable Edad

La variable Edad es cuantitativa discreta porque representa la edad de las personas en años. Su promedio permite conocer la edad media de los casos, mientras que la mediana permite identificar el valor central de la distribución.

La comparación entre promedio y mediana también puede ayudar a identificar si existen diferencias importantes en la distribución de las edades.

```python
data['Edad'].describe()

print("Edad promedio:", data['Edad'].mean())
print("Edad mediana:", data['Edad'].median())
print("Edad mínima:", data['Edad'].min())
print("Edad máxima:", data['Edad'].max())
```

Análisis de variables categóricas

```python
variables_categoricas = data.select_dtypes(
    include=['object', 'category']
).columns

print("Variables categóricas:")
print(variables_categoricas)

for columna in variables_categoricas:
    print("\n====================================")
    print("VARIABLE:", columna)
    print("====================================")
    print(data[columna].value_counts(dropna=False))
```

```python
resumen = pd.DataFrame({
    'Variable': data.columns,
    'Tipo de dato': data.dtypes.astype(str).values,
    'Valores no nulos': data.notnull().sum().values,
    'Valores faltantes': data.isnull().sum().values,
    'Valores únicos': [
        data[col].nunique(dropna=False) for col in data.columns
    ]
})

resumen
```

Conclusion

- El análisis descriptivo realizado sobre el conjunto de datos de casos de COVID-19 en Colombia permitió conocer las principales características de las variables y la forma en que se distribuyen los registros. Se encontraron variables de diferentes naturalezas: cuantitativas, como la edad y los identificadores numéricos, cualitativas, como sexo, estado, atención, tipo, departamento y pertenencia étnica, y variables relacionadas con fechas, como la fecha de notificación, diagnóstico, recuperación y muerte.

```python
from ydata_profiling import ProfileReport
```

```python
# Crear una muestra de 50.000 registros
data_muestra = data.sample(
    n=min(50000, len(data)),
    random_state=42
)
```

```python
profile = ProfileReport(
    data_muestra,
    title="Perfilado de datos - COVID-19 en Colombia",
    explorative=True
)

profile
```

CONCLUSIONES

- El perfilado realizado al conjunto de datos de COVID-19 en Colombia permitió identificar diferentes características y posibles anomalías relacionadas con la calidad de los datos. En primer lugar, se encontraron valores faltantes en varias variables. Estos valores no necesariamente representan errores, ya que algunas variables dependen de la evolución del caso. Por ejemplo, las fechas de muerte o recuperación pueden no estar registradas cuando el evento no ocurrió o no fue reportado.

- Finalmente, el perfilado permite concluir que, aunque el conjunto de datos contiene información suficiente para realizar análisis descriptivos y estadísticos, es necesario realizar previamente un proceso de limpieza y validación de los datos. Las principales situaciones que deben tenerse en cuenta son los valores faltantes, posibles duplicados, valores atípicos, inconsistencias en variables categóricas y posibles errores en las fechas. Estas anomalías deben ser documentadas y tratadas cuidadosamente para evitar modificar información válida y garantizar que los análisis posteriores sean confiables.
